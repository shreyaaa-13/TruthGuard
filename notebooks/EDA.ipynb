# Fake News Detection - Exploratory Data Analysis
## Analyzing the Fake and Real News Dataset

This notebook performs comprehensive exploratory data analysis on the fake news dataset.

```python
# Import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from wordcloud import WordCloud
import warnings
warnings.filterwarnings('ignore')

# Set style
sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (12, 6)

print("Libraries imported successfully!")
```

## 1. Load the Data

```python
# Load datasets
fake_df = pd.read_csv('../data/Fake.csv')
real_df = pd.read_csv('../data/True.csv')

# Add labels
fake_df['label'] = 'FAKE'
real_df['label'] = 'REAL'

print(f"Fake news articles: {len(fake_df)}")
print(f"Real news articles: {len(real_df)}")
print(f"Total articles: {len(fake_df) + len(real_df)}")
```

## 2. Data Overview

```python
# Display first few rows of fake news
print("=" * 70)
print("FAKE NEWS SAMPLE")
print("=" * 70)
fake_df.head()
```

```python
# Display first few rows of real news
print("=" * 70)
print("REAL NEWS SAMPLE")
print("=" * 70)
real_df.head()
```

```python
# Data info
print("Fake News Dataset Info:")
print(fake_df.info())
print("\nReal News Dataset Info:")
print(real_df.info())
```

## 3. Missing Values Analysis

```python
# Check missing values
print("Missing values in Fake News:")
print(fake_df.isnull().sum())
print("\nMissing values in Real News:")
print(real_df.isnull().sum())
```

## 4. Text Length Analysis

```python
# Calculate text lengths
fake_df['text_length'] = fake_df['text'].str.len()
real_df['text_length'] = real_df['text'].str.len()

fake_df['title_length'] = fake_df['title'].str.len()
real_df['title_length'] = real_df['title'].str.len()

# Statistics
print("Text Length Statistics:")
print("\nFake News:")
print(fake_df['text_length'].describe())
print("\nReal News:")
print(real_df['text_length'].describe())
```

```python
# Visualize text length distribution
fig, axes = plt.subplots(1, 2, figsize=(15, 5))

axes[0].hist(fake_df['text_length'], bins=50, alpha=0.7, color='red', label='Fake')
axes[0].hist(real_df['text_length'], bins=50, alpha=0.7, color='green', label='Real')
axes[0].set_xlabel('Text Length')
axes[0].set_ylabel('Frequency')
axes[0].set_title('Distribution of Article Text Length')
axes[0].legend()

axes[1].hist(fake_df['title_length'], bins=50, alpha=0.7, color='red', label='Fake')
axes[1].hist(real_df['title_length'], bins=50, alpha=0.7, color='green', label='Real')
axes[1].set_xlabel('Title Length')
axes[1].set_ylabel('Frequency')
axes[1].set_title('Distribution of Title Length')
axes[1].legend()

plt.tight_layout()
plt.show()
```

## 5. Word Count Analysis

```python
# Calculate word counts
fake_df['word_count'] = fake_df['text'].str.split().str.len()
real_df['word_count'] = real_df['text'].str.split().str.len()

# Visualize
fig, ax = plt.subplots(figsize=(12, 6))

ax.hist(fake_df['word_count'], bins=50, alpha=0.7, color='red', label='Fake News')
ax.hist(real_df['word_count'], bins=50, alpha=0.7, color='green', label='Real News')
ax.set_xlabel('Word Count')
ax.set_ylabel('Frequency')
ax.set_title('Distribution of Word Count in Articles')
ax.legend()

plt.show()

print(f"Average word count in fake news: {fake_df['word_count'].mean():.2f}")
print(f"Average word count in real news: {real_df['word_count'].mean():.2f}")
```

## 6. Word Clouds

```python
# Create word clouds
fig, axes = plt.subplots(1, 2, figsize=(20, 8))

# Fake news word cloud
fake_text = ' '.join(fake_df['text'].astype(str))
fake_wordcloud = WordCloud(width=800, height=400, background_color='white', 
                           colormap='Reds').generate(fake_text)

axes[0].imshow(fake_wordcloud, interpolation='bilinear')
axes[0].axis('off')
axes[0].set_title('Fake News Word Cloud', fontsize=20, fontweight='bold')

# Real news word cloud
real_text = ' '.join(real_df['text'].astype(str))
real_wordcloud = WordCloud(width=800, height=400, background_color='white',
                           colormap='Greens').generate(real_text)

axes[1].imshow(real_wordcloud, interpolation='bilinear')
axes[1].axis('off')
axes[1].set_title('Real News Word Cloud', fontsize=20, fontweight='bold')

plt.tight_layout()
plt.show()
```

## 7. Subject/Topic Analysis

```python
# Check if 'subject' column exists
if 'subject' in fake_df.columns:
    print("Fake News Topics:")
    print(fake_df['subject'].value_counts())
    
    print("\nReal News Topics:")
    print(real_df['subject'].value_counts())
    
    # Visualize
    fig, axes = plt.subplots(1, 2, figsize=(15, 6))
    
    fake_df['subject'].value_counts().plot(kind='bar', ax=axes[0], color='red', alpha=0.7)
    axes[0].set_title('Fake News Topics Distribution')
    axes[0].set_xlabel('Topic')
    axes[0].set_ylabel('Count')
    axes[0].tick_params(axis='x', rotation=45)
    
    real_df['subject'].value_counts().plot(kind='bar', ax=axes[1], color='green', alpha=0.7)
    axes[1].set_title('Real News Topics Distribution')
    axes[1].set_xlabel('Topic')
    axes[1].set_ylabel('Count')
    axes[1].tick_params(axis='x', rotation=45)
    
    plt.tight_layout()
    plt.show()
```

## 8. Combined Dataset Analysis

```python
# Combine datasets
df = pd.concat([fake_df, real_df], ignore_index=True)

print(f"Total dataset size: {len(df)}")
print(f"\nLabel distribution:")
print(df['label'].value_counts())

# Visualize label distribution
fig, ax = plt.subplots(figsize=(8, 6))
df['label'].value_counts().plot(kind='bar', ax=ax, color=['red', 'green'], alpha=0.7)
ax.set_title('Distribution of Fake vs Real News', fontsize=16, fontweight='bold')
ax.set_xlabel('Label', fontsize=12)
ax.set_ylabel('Count', fontsize=12)
ax.tick_params(axis='x', rotation=0)

# Add value labels on bars
for i, v in enumerate(df['label'].value_counts()):
    ax.text(i, v + 100, str(v), ha='center', fontweight='bold')

plt.tight_layout()
plt.show()
```
